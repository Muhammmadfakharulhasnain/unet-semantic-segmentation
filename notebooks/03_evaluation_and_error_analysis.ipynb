**Goal:** interpret model performance instead of reporting one accuracy number.

The project evaluates the held-out test set using a confusion matrix and class-wise overlap metrics. It also saves visual comparisons: input, ground truth, prediction and pixel-error map.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent))
import pandas as pd
import matplotlib.pyplot as plt

results_path = Path("../results/tables/experiment_comparison.csv")
if results_path.exists():
    results = pd.read_csv(results_path)
    display(results)
else:
    print("Run the experiments before interpreting test results.")
```

## 1. Metrics

- IoU = TP / (TP + FP + FN)
- Dice = 2TP / (2TP + FP + FN)
- Precision = TP / (TP + FP)
- Recall = TP / (TP + FN)
- Pixel accuracy = all correctly classified pixels / all pixels

Mean IoU and mean Dice average across the three classes. Foreground mean IoU in this project averages pet and boundary, excluding the background class. State this definition whenever reporting it.

```python
if results_path.exists():
    metric_cols = [c for c in results.columns if c.startswith("test_") and
                   any(k in c for k in ["mean_iou", "foreground_mean_iou",
                                        "mean_dice", "pixel_accuracy",
                                        "macro_precision", "macro_recall"])]
    display(results[["experiment", "augmentation", "loss"] + metric_cols])
```

## 2. Training curves

Each run produces a loss curve and mean-IoU curve for training and validation. A widening gap between training and validation performance can indicate overfitting. A flat curve can indicate underfitting, insufficient training, or optimization difficulty.

```python
from IPython.display import Image, display
fig_dir = Path("../results/figures")
if fig_dir.exists():
    for path in sorted(fig_dir.glob("*_curves.png")):
        print(path.name)
        display(Image(filename=str(path)))
else:
    print("No figures yet. Complete training first.")
```
